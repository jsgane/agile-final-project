```python
# Configuration et imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from datetime import timedelta
import json
import pandas as pd
from pandas.api.types import is_string_dtype
from pandas.api.types import is_numeric_dtype
```

```python
#loading the csv file and displaying the first five rows
CSV_PATH = "employees_west_africa.csv"

df = pd.read_csv(CSV_PATH)

print(df.head())
```

```python
# removing duplicate rows
df.duplicated().sum()
df = df.drop_duplicates()
# printing the results
print(df.head())
```

```python
# Handle missing values 
print(df)
print(df.isnull().sum())
print(df.shape)
```

```python
# For Categorial values we are filling with 'Unknown'
df['name'] = df['name'].fillna('Unknown')
df['department'] = df['department'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
# For numerical with the mean
df['age'] = df['age'].fillna(df['age'].mean())
df['salary'] = df['salary'].fillna(df['salary'].mean())

print(df)
```

```python
print(type(df))
```

```python
# Standardizing the gender column 
df.loc[df['gender'] == 'M', 'gender'] = 'Male'
df.loc[df['gender'] == 'male', 'gender'] = 'Male'
df.loc[df['gender'] == 'F', 'gender'] = 'Female'
df.loc[df['gender'] == 'FEMALE', 'gender'] = 'Female'
df.loc[df['gender'] == 'female', 'gender'] = 'Female'
print(df)

```

```python
# Convert the hire_date column to a valid datetime format
df['hire_date'] = df['hire_date'].str.replace('/', '-', regex=False)
df['hire_date'] = pd.to_datetime(df['hire_date'], errors = 'coerce')
print(df)
```

```python
#Removing any extra spaces in column names and text values
df.columns = df.columns.str.strip()
print(df)
```

```python
#Creating a new column years_worked showing the number of years since the employee was hired.
df['years_worked'] = round((pd.Timestamp.today() - df['hire_date']).dt.days / 365, 2)
df['years_worked'] = df['years_worked'].fillna(0)
print(df)
```

```python
#Computing the average salary per department
df_avg_sal = df.groupby('name', as_index = False)['salary'].sum()
print(df_avg_sal)
```

```python
#Identifying the top 5 highest-paid employees
df_top5_emp = df.groupby(['employee_id','name'], as_index = False)['salary'].sum().sort_values(by = 'salary', ascending = False)
print(df_top5_emp.head())
```

```python
#Detecting employees with inconsistent ages (e.g., under 18 or over 65).
df_emp_inc = df[(df['age'] < 18) | (df['age'] > 65)]
print(df_emp_inc)
```

```python
# Exporting cleaned data
emp_clean_csv = 'cleaned_employees_west_africa.csv'
df.to_csv(emp_clean_csv, index=False)
print('Fichier pré-traité sauvegardé :', emp_clean_csv)
```
